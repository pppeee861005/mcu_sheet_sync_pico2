## `requests` - 倫敦希斯洛機場每月溫度、雨量

* 使用 `urllib` 程式庫

```python
from urllib.request import urlopen, Request
from urllib.error import HTTPError
from urllib.error import URLError

# url = ("http://www.metoffice.gov.uk/pub/data/weather/uk/climate/"
#        "stationdata/heathrowdata.txt")
url = "https://google.com/not-exists"

with urlopen(url) as conn:
    print(conn.getheaders())
    data = conn.read().decode()

print(data)
```

* 捕捉異常

```python
try:
    conn = urlopen(url)
except HTTPError as e1:
    print('HTTP error:', e1)
except URLError as e2:
    print('Server not found!:', e2)
else:
    print(conn.read())
```

* 使用 `requests` 程式庫

```python
import requests
from requests.exceptions import HTTPError
```

```python
# url = ("http://www.metoffice.gov.uk/pub/data/weather/uk/climate/"
#        "stationdata/heathrowdata.txt")
url = "https://google.com/not-exists"

response = requests.get(url)
response.raise_for_status()

print(response.text)
print(response.status_code)
```

* 捕捉異常

```python
try:
    response = requests.get(url)
    response.raise_for_status()
except HTTPError as http_err:
    print(f'HTTP error occurred: {http_err}')  # Python 3.6
except Exception as err:
    print(f'Other error occurred: {err}')  # Python 3.6
else:
    print('Success!')
```

```python
url = 'http://www.google.com/404'

try:
    r = requests.get(url, timeout=3)
    r.raise_for_status()
except requests.exceptions.RequestException as ex1:
    print("Http 請求錯誤: " + str(ex1))
except requests.exceptions.HTTPError as ex2:
    print("Http 回應錯誤: " + str(ex2))
except requests.exceptions.ConnectionError as ex3:
    print("網路連線錯誤: " + str(ex3))
except requests.exceptions.Timeout as ex4:
    print("Timeout 錯誤: " + str(ex4))
```

## Oauth - 存取 Google 試算表

* 驗證 (固定寫法)

```python
import gspread
from pathlib import Path
from oauth2client.service_account import ServiceAccountCredentials as sac

# WORK = Path(__file__).resolve().parent
WORK = Path('d:/code秋季115/week_09')
oauth = WORK / 'google_auth.json'
# oauth = 'google_auth.json'
scope = ['https://spreadsheets.google.com/feeds',
         'https://www.googleapis.com/auth/drive']

# 建立憑證
cr = sac.from_json_keyfile_name(oauth, scope)  # 請自行修改檔名
gs = gspread.authorize(cr)
```

* 開啟工作表

```python
sh = gs.open('智慧農場試算表')
# wks = sh.sheet1
wks = sh.worksheet('銷售報表')
```

* 更新儲存格

```python
wks.update_acell('C2', 'Green Tea')
wks.update_cell(4, 3, 'Black Tea')   # 第 2 列，第 3 欄儲存格
```

* 寫入整列

```python
values = ['Name', 'Year', 'Product', 'Price', 'Quantity', 'Revenue', 'Location']
wks.insert_row(values, 1)
```

* 讀取儲存格

```python
print(wks.acell('A1').value)
print(wks.cell(1, 1).value)
print(wks.row_values(1))
print(wks.col_values(1))
print(wks.get_all_values())
```

* 保留第一列，其餘刪除
